# Axisymmetric LNS with Floating Disk — Visualisation

Adapts the pure-fluid axisymmetric notebook of folder 11 to the
self-consistent floating-disk problem of folder 12.
The body is a flat-bottomed cylinder of radius $a$, mass $m$, released from
$z_b(0) = 0$ at rest. With negligible surface tension the expected hydrostatic
equilibrium is $|z_b^{\text{eq}}| = M/(\pi a^2)$ (3D Archimedes).

```python
from floating_object_axi import (
    FloatingObjectSolverAxi, SimConfig,
    PhysicalParams, BodyParams, NumericalParams,
)

# Viscous + low surface tension for clean settling toward Archimedes equilibrium
config = SimConfig(
    phys=PhysicalParams(L_cm=50.0, D_cm=20.0, nu=2.0, sigma=1.0),
    body=BodyParams(radius_cm=2.5, mass_g=7.0,
                    initial_z_cm=0.0, initial_vz_cm_s=0.0),
    num=NumericalParams(nr=200, nz=40, nt=3000, total_time_s=3.0),
)
solver = FloatingObjectSolverAxi(config)
result = solver.run()
```

## Plot 1: body position trace + radial surface snapshots


```python
import matplotlib.pyplot as plt
import numpy as np

uL = config.num.unit_length
ut = config.num.unit_time
r_cm = result["r"] * uL
t = result["time"]
eta_arr = np.array(result["eta_history"]) * uL * 10            # mm
ps_arr = np.array(result["pressure_history"])                  # dim'less
zb_arr = result["z_b"] * uL * 10                               # mm
vb_arr = result["v_b"] * uL / ut                               # cm/s
zb_eq_mm = result["zb_eq_predicted"] * uL * 10
a_cm = solver.a_obj * uL

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

ax = axes[0]
ax.plot(t, zb_arr, "b-", lw=2, label="$z_b(t)$")
ax.axhline(zb_eq_mm, color="r", ls="--", lw=1.5,
           label=f"$-M/(\\pi a^2)$ = {zb_eq_mm:.3f} mm")
ax.axhline(0, color="k", ls="-", lw=0.5, alpha=0.5)
ax.set(xlabel="Time (s)", ylabel="z (mm)", title="Body bottom position $z_b(t)$")
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

ax = axes[1]
n_snap = 7
idx = np.linspace(0, len(t) - 1, n_snap, dtype=int)
colors = plt.cm.viridis(np.linspace(0, 1, n_snap))
for i, c in zip(idx, colors):
    ax.plot(r_cm, eta_arr[i], color=c, lw=1.5, label=f"t={t[i]:.2f}s")
ax.axvline(a_cm, color="gray", ls=":", lw=1.5, label=f"body edge r=a={a_cm:.2f} cm")
ax.axhline(0, color="k", ls="-", lw=0.5, alpha=0.5)
ax.set(xlabel="r (cm)", ylabel="$\\eta$ (mm)",
       title="Radial snapshots of $\\eta(r, t)$ (body region pinned to $z_b$)")
ax.legend(fontsize=7, ncol=2)
ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()
```

## Plot 2: space-time diagram of $\eta(r, t)$ with body-edge marker


```python
fig, ax = plt.subplots(figsize=(10, 5))
vmax = max(1e-6, np.percentile(np.abs(eta_arr), 99))
im = ax.imshow(eta_arr, aspect="auto", origin="lower",
               extent=[r_cm[0], r_cm[-1], t[0], t[-1]],
               cmap="RdBu_r", vmin=-vmax, vmax=vmax)
ax.axvline(a_cm, color="black", ls="--", lw=1.5, label=f"body edge $r=a$")
ax.set(xlabel="r (cm)", ylabel="Time (s)",
       title="$\\eta(r, t)$ space-time diagram")
ax.legend(loc="upper right", fontsize=9)
plt.colorbar(im, ax=ax, label="$\\eta$ (mm)")
plt.show()
```

## Plot 3: top-down (rotated) view with disk overlaid


```python
def render_topdown(eta_radial, r_cm, n_pixels=300):
    '''Reflect a radial profile into a 2D top-down image.'''
    R = r_cm[-1]
    x = np.linspace(-R, R, n_pixels)
    y = np.linspace(-R, R, n_pixels)
    X, Y = np.meshgrid(x, y)
    rho = np.sqrt(X**2 + Y**2)
    eta_2d = np.interp(rho, r_cm, eta_radial, right=0.0)
    return X, Y, eta_2d

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
times_to_show = [len(t) // 4, len(t) // 2, 3 * len(t) // 4]
vmax = max(1e-6, np.percentile(np.abs(eta_arr), 99))
for ax, idx in zip(axes, times_to_show):
    X, Y, eta_2d = render_topdown(eta_arr[idx], r_cm)
    im = ax.imshow(eta_2d, origin="lower",
                   extent=[X.min(), X.max(), Y.min(), Y.max()],
                   cmap="RdBu_r", vmin=-vmax, vmax=vmax)
    disk_outline = plt.Circle((0, 0), a_cm, fill=False,
                              color="black", lw=1.5, ls="--")
    ax.add_patch(disk_outline)
    ax.set(xlabel="x (cm)", ylabel="y (cm)",
           title=f"$\\eta$ at t={t[idx]:.3f}s,  $z_b$={zb_arr[idx]:.2f} mm")
    ax.set_aspect("equal")
    plt.colorbar(im, ax=ax, label="$\\eta$ (mm)", fraction=0.046)
fig.tight_layout()
plt.show()
```

## Plot 4: GIF — radial profile (with body) + top-down view


```python
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.patches import Rectangle
import os

os.makedirs("Outputs", exist_ok=True)

gif_duration_seconds = 8.0
n_frames = len(t)
fps = max(1, int(round(n_frames / gif_duration_seconds)))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax1 = axes[0]
y_min = min(float(eta_arr.min()), float(zb_arr.min()), float(zb_eq_mm)) - 1.0
y_max = max(float(eta_arr.max()), 1.0)
line_eta, = ax1.plot(r_cm, eta_arr[0], "b-", lw=2, label="$\\eta(r,t)$")
body_height_mm = 5.0
body_patch = Rectangle((0, zb_arr[0]), a_cm, body_height_mm,
                       facecolor="gray", alpha=0.5, edgecolor="black", label="body")
ax1.add_patch(body_patch)
ax1.axhline(zb_eq_mm, color="r", ls="--", lw=1, alpha=0.6, label="hydrostatic")
ax1.axhline(0, color="k", ls="-", lw=0.5)
ax1.set(xlim=[r_cm[0], r_cm[-1]], ylim=[y_min, y_max + body_height_mm + 1],
        xlabel="r (cm)", ylabel="z (mm)")
title1 = ax1.set_title("t = 0.000 s")
ax1.legend(loc="lower right", fontsize=8)

ax2 = axes[1]
X, Y, eta_2d = render_topdown(eta_arr[0], r_cm)
vmax = max(1e-6, np.percentile(np.abs(eta_arr), 99))
im = ax2.imshow(eta_2d, origin="lower", extent=[X.min(), X.max(), Y.min(), Y.max()],
                cmap="RdBu_r", vmin=-vmax, vmax=vmax)
disk_outline = plt.Circle((0, 0), a_cm, fill=False,
                          color="black", lw=1.5, ls="--")
ax2.add_patch(disk_outline)
ax2.set(xlabel="x (cm)", ylabel="y (cm)")
ax2.set_aspect("equal")
title2 = ax2.set_title("Top-down view: t = 0.000 s")
plt.colorbar(im, ax=ax2, label="$\\eta$ (mm)", fraction=0.046)

def update(i):
    line_eta.set_ydata(eta_arr[i])
    body_patch.set_y(zb_arr[i])
    _, _, eta_2d_i = render_topdown(eta_arr[i], r_cm)
    im.set_data(eta_2d_i)
    title1.set_text(f"t = {t[i]:.3f} s,  $z_b$ = {zb_arr[i]:+.3f} mm")
    title2.set_text(f"Top-down: t = {t[i]:.3f} s")
    return line_eta, body_patch, im

anim = FuncAnimation(fig, update, frames=n_frames, interval=40, blit=False)
anim.save("./Outputs/floating_axi.gif", writer=PillowWriter(fps=fps), dpi=120)
plt.show()
print(f"Saved Outputs/floating_axi.gif  (fps={fps}, duration ≈ {n_frames/fps:.1f} s)")
```

## Diagnostics: body dynamics, force balance, energy


```python
fig, axes = plt.subplots(2, 2, figsize=(13, 8))

ax = axes[0, 0]
ax.plot(t, zb_arr, "b-", lw=2, label="$z_b(t)$")
ax.axhline(zb_eq_mm, color="r", ls="--", lw=1.5,
           label=f"$-M/(\\pi a^2)$ = {zb_eq_mm:.3f} mm")
ax.set(xlabel="Time (s)", ylabel="z (mm)", title="Body position $z_b(t)$")
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

ax = axes[0, 1]
ax.plot(t, vb_arr, "g-", lw=2)
ax.axhline(0, color="k", ls="-", lw=0.5)
ax.set(xlabel="Time (s)", ylabel="$v_b$ (cm/s)", title="Body velocity $v_b(t)$")
ax.grid(alpha=0.3)

ax = axes[1, 0]
F_body = result["body_force"]
weight = solver.M / solver.Fr
ax.plot(t, F_body, "b-", lw=2, label="$\\int p_s\\, 2\\pi r\\, dr$")
ax.axhline(weight, color="r", ls="--", lw=1.5,
           label=f"weight $M/Fr$ = {weight:.3f}")
ax.set(xlabel="Time (s)", ylabel="Force (dim'less)",
       title="Integrated traction vs body weight")
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

ax = axes[1, 1]
E = result["energy"]
ax.plot(t, E, "r-", lw=2)
ax.set(xlabel="Time (s)", ylabel="Energy (dim'less)",
       title="Coupled energy (body + fluid PE)")
ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()

n_late = max(1, len(zb_arr) // 5)
print(f"Final z_b: {zb_arr[-1]:+.4f} mm")
print(f"Predicted z_b^eq (continuous): {zb_eq_mm:+.4f} mm")
print(f"Late-time <z_b>: {np.mean(zb_arr[-n_late:]):+.4f} "
      f"+/- {np.std(zb_arr[-n_late:]):.4f} mm")
print(f"Final integrated force / weight: {F_body[-1]/weight:.4f}")
```

## 3D rendered GIF (with disk visible above the wavy surface)


```python
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.colors import LightSource
import os

os.makedirs("Outputs", exist_ok=True)

n_pixels = 180
zoom = 0.4              # render only inner part (body is small relative to L)
gif_duration_seconds = 8.0
elev, azim = 28, 45
exaggeration = 10

R_max = float(r_cm[-1]) * zoom
xs = np.linspace(-R_max, R_max, n_pixels)
ys = np.linspace(-R_max, R_max, n_pixels)
Xg, Yg = np.meshgrid(xs, ys)
rho = np.sqrt(Xg**2 + Yg**2)

def eta_2d_for_frame(idx):
    eta_2d = np.interp(rho, r_cm, eta_arr[idx], right=0.0)
    eta_2d[rho > R_max] = np.nan
    return eta_2d

# Disk geometry (for plotting on top of the surface)
theta = np.linspace(0, 2 * np.pi, 60)
disk_x = a_cm * np.cos(theta)
disk_y = a_cm * np.sin(theta)

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

z_extent = float(np.nanmax(np.abs(eta_arr))) * exaggeration
ls = LightSource(azdeg=315, altdeg=45)
cmap = plt.cm.coolwarm

eta_2d_0 = eta_2d_for_frame(0)
Z0 = eta_2d_0 * exaggeration
surf = [ax.plot_surface(Xg, Yg, Z0, rcount=80, ccount=80,
                        cmap=cmap, lightsource=ls,
                        vmin=-z_extent, vmax=z_extent,
                        antialiased=True, edgecolor="none")]
# Disk top edge — drawn at z_b * exaggeration + small thickness
disk_line = [ax.plot(disk_x, disk_y,
                     zs=zb_arr[0] * exaggeration,
                     color="black", lw=2)[0]]

ax.set_box_aspect((1, 1, 0.45))
ax.set_xlabel("x (cm)")
ax.set_ylabel("y (cm)")
ax.set_zlabel(f"$\\eta\\times{exaggeration}$ (mm)")
ax.set_zlim(-1.05 * z_extent, 1.05 * z_extent)
ax.view_init(elev=elev, azim=azim)
title = ax.set_title("t = 0.000 s,  $z_b$ = 0.000 mm")

ax.xaxis.pane.set_alpha(0.05)
ax.yaxis.pane.set_alpha(0.05)
ax.zaxis.pane.set_alpha(0.05)

def update3d(idx):
    surf[0].remove()
    disk_line[0].remove()
    eta_2d_i = eta_2d_for_frame(idx)
    Z = eta_2d_i * exaggeration
    surf[0] = ax.plot_surface(Xg, Yg, Z, rcount=80, ccount=80,
                              cmap=cmap, lightsource=ls,
                              vmin=-z_extent, vmax=z_extent,
                              antialiased=True, edgecolor="none")
    disk_line[0] = ax.plot(disk_x, disk_y,
                           zs=zb_arr[idx] * exaggeration,
                           color="black", lw=2)[0]
    title.set_text(f"t = {t[idx]:.3f} s,  $z_b$ = {zb_arr[idx]:+.3f} mm")
    return surf[0], disk_line[0]

n_frames = len(t)
fps = max(1, int(round(n_frames / gif_duration_seconds)))

anim = FuncAnimation(fig, update3d, frames=n_frames, interval=40, blit=False)
anim.save("./Outputs/floating_axi_3d.gif",
          writer=PillowWriter(fps=fps), dpi=110)
plt.show()
print(f"Saved Outputs/floating_axi_3d.gif  ({fps} fps, "
      f"≈ {n_frames/fps:.1f} s playback, exaggeration ×{exaggeration})")
```
